# Условный экстремум (визуализация)

```python
import numpy as np
import plotly.graph_objects as go
```

Рассмотрим задачу
\begin{gather}
		\min(x^2+2y^2) \\ s.t.\;3x+2y=22
\end{gather}

```python
X = np.arange(start=-20, stop=50, step=0.2)
Y = np.arange(start=-20, stop=50, step=0.2)
X, Y = np.meshgrid(X, Y)
Z = X**2+2*Y**2

X_constr = np.arange(start=-20, stop=20, step=0.2)
Y_constr = 22-1.5*X_constr
Z_constr = np.full(len(X_constr), 0)
Z_fun = X_constr**2+2*Y_constr**2

# fig = px.line_3d(x=X_constr, y=Y_constr, z=Z_constr)
fig = go.Figure(data=[ go.Surface(x=X, y=Y, z=Z, opacity=0.7)])
fig.add_traces(go.Scatter3d(x=X_constr, y=Y_constr, z=Z_constr, marker_size=1))
fig.add_traces(go.Scatter3d(x=X_constr, y=Y_constr, z=Z_fun, marker_size=1))
fig.show()
```
